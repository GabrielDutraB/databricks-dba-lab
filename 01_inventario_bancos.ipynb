```python
from pyspark.sql import Row

dados = [
    Row(servidor="SQL01", banco="Vendas", tamanho_gb=120.5, ultimo_backup="2026-10-12"),
    Row(servidor="SQL01", banco="Estoque", tamanho_gb=45.2, ultimo_backup="2026-10-12"),
    Row(servidor="SQL02", banco="Financeiro", tamanho_gb=210.8, ultimo_backup="2026-10-10"),
    Row(servidor="SQL02", banco="RH", tamanho_gb=8.3, ultimo_backup="2026-10-12"),
]

df = spark.createDataFrame(dados)
df.createOrReplaceTempView("bancos")
display(df)
```

```python
%sql
SELECT servidor,
       COUNT(*) AS qtd_bancos,
       ROUND(SUM(tamanho_gb), 1) AS total_gb
FROM bancos
GROUP BY servidor
ORDER BY total_gb DESC
```
